# synchronous programming


```python
import time
```

```python
def fetch_weather():
    print("Fetching weather data...")   
    time.sleep(4)  # Simulate a delay in fetching data
    print("Weather data fetched.")

def fetch_news():
    print("Fetching news data...")
    time.sleep(2)  # Simulate a delay in fetching data
    print("News data fetched.")



def main():
    start_time = time.time()
    
    fetch_weather()
    fetch_news()
    
    end_time = time.time()
    print(f"Total time taken: {end_time - start_time:.2f} seconds") 
```

```python
main()
```

# Async

```python
import asyncio
import time
```

```python
async def fetch_weather():
    print("Fetching weather data...")   
    await asyncio.sleep(4)  # Simulate a delay in fetching data

    print("Weather data fetched.")

async def fetch_news():
    print("Fetching news data...")
    await asyncio.sleep(2)  # Simulate a delay in fetching data
    print("News data fetched.")



async def main():
    start_time = time.time()
    
    await asyncio.gather(fetch_weather(), fetch_news())
    
    end_time = time.time()
    print(f"Total time taken: {end_time - start_time:.2f} seconds") 
```

```python
await main()
```
